```python
#importing pandas
import pandas as pd
```

```python
#define file location
file_path = '/content/Salary_Data.csv'
#load data
salary = pd.read_csv(file_path)
salary.head(2)
```

```python
#count missing or null values
salary.isnull().sum()
```

```python
#prepare data
X = salary[['YearsExperience']]
y = salary[['Salary']]
```

```python
salary[['Salary']].values
```

```python
#analyis
import matplotlib.pyplot as plt
plt.scatter(X,y,color='red',marker='+')
plt.xlabel('Years of Experience')
plt.ylabel('Salary')
plt.title('Analysis of Linear Relationship',color='Green')
```

```python
#import algorithm --> sckit learn --> sklearn

from sklearn.linear_model import LinearRegression
model = LinearRegression()
```

```python
model
```

```python
#As it is class we need to make constructor
model.fit(X,y)
```
